```python
import matplotlib.pyplot as plt
import numpy as np


# 1. Definição da Função Ackley 2D
def ackley(x, y):
    termo1 = -20 * np.exp(-0.2 * np.sqrt(0.5 * (x**2 + y**2)))
    termo2 = -np.exp(0.5 * (np.cos(2 * np.pi * x) + np.cos(2 * np.pi * y)))
    constante = 20 + np.e

    return termo1 + termo2 + constante


# 2. Configuração dos Parâmetros do Algoritmo
NUM_TENTATIVAS = 2000
LIXO_MIN, LIXO_MAX = -32.768, 32.768

melhor_custo = float("inf")
melhor_x, melhor_y = None, None
historico = []

# 3. Execução da Busca Aleatória (Random Search)
for _ in range(NUM_TENTATIVAS):
    # Gera coordenadas aleatórias no intervalo [-32.768, 32.768]
    x = np.random.uniform(LIXO_MIN, LIXO_MAX)
    y = np.random.uniform(LIXO_MIN, LIXO_MAX)

    # Avalia o ponto atual
    custo = ackley(x, y)

    # Atualiza o melhor resultado até o momento
    if custo < melhor_custo:
        melhor_custo = custo
        melhor_x = x
        melhor_y = y

    # Registra o progresso do menor custo para o gráfico
    historico.append(melhor_custo)

# 4. Exibição dos Resultados
print("=== RESULTADOS FINAIS ===")
print(f"Melhor valor encontrado (f(x, y)): {melhor_custo:.6f}")
print(f"Melhores coordenadas: x = {melhor_x:.6f}, y = {melhor_y:.6f}")

# 5. Gráfico de Convergência
plt.figure(figsize=(10, 5))
plt.plot(historico, color="blue", linewidth=2, label="Melhor Custo")
plt.title("Gráfico de Convergência - Função Ackley (Random Search)")
plt.xlabel("Tentativas (Iterações)")
plt.ylabel("Menor Custo Encontrado")
plt.grid(True, linestyle="--", alpha=0.7)
plt.legend()
plt.show()
```
