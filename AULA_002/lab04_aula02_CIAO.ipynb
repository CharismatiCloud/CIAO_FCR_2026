```python
Atividade 4 — DO ZERO: Modelagem de um Problema Real

1 - Objetivo da atividade: aplicar os conceitos da aula a um problema que faca sentido para ele, consolidando a aprendizagem.
Tarefa:
Escolha um problema real do seu cotidiano, da sua area de trabalho ou de interesse
(exemplos: montar a grade de horarios da faculdade, escolher quais pedidos atender em um dia, rotear entregas, alocar pessoas em projetos, montar a lista de compras com orcamento limitado, etc.).

2 - Entregar um relatorio curto de 1–2 paginas + codigo.
Descricao clara do problema em linguagem natural.
Modelagem formal contendo:
•	O que e uma solucao (como voce representa uma solucao candidata).
•	Qual e o espaco de busca (quantas solucoes existem aproximadamente?).
•	Qual e a funcao objetivo (o que voce quer maximizar ou minimizar?).
•	Quais sao as restricoes (o que torna uma solucao invalida?).

3 - Classificacao: voce considera este problema “facil” (existe metodo rapido e exato) ou “dificil” (parece NP-dificil)? Justifique de forma simples.
Um codigo Python que:
•	Gera pelo menos uma solucao aleatoria (factivel ou nao).
•	Calcula o valor da funcao objetivo dessa solucao.
•	Verifica se a solucao respeita as restricoes.

4 - Sugestoes de problemas para quem estiver sem ideia
•	Escolher quais materias cursar no proximo semestre (respeitando pre-requisitos e carga horaria maxima).
•	Montar a lista de compras do mes com orcamento limitado, maximizando “utilidade”.
•	Alocar alunos em grupos de trabalho com restricoes de preferencia e equilibrio de habilidades.
•	Definir a ordem de atendimento de pacientes em um pronto-socorro (minimizar tempo medio de espera ou gravidade acumulada).
•	Escolher quais filmes/series assistir no fim de semana maximizando satisfacao com limite de tempo.

5 - Criterios de avaliacao desta atividade
•	Clareza da modelagem (espaco de busca, objetivo e restricoes bem definidos).
•	Coerencia entre a descricao e o codigo.
•	Qualidade da justificativa sobre a dificuldade do problema.
•	Criatividade e relevancia do problema escolhido (nao e o mais importante, mas valoriza).


1. Descrição do problema

O problema escolhido é a montagem de uma lista de compras mensal com orçamento limitado. A ideia é escolher quais produtos comprar de forma que o dinheiro disponível seja respeitado e, ao mesmo tempo, a utilidade total dos produtos escolhidos seja maximizada.
Por exemplo, uma pessoa possui um orçamento de R$ 300 para realizar suas compras do mês. Cada produto possui um preço e um valor de utilidade, que representa o quanto aquele produto é importante ou útil para a pessoa. O objetivo é escolher os produtos que proporcionem a maior utilidade possível sem ultrapassar o orçamento.
Esse problema representa uma situação comum do cotidiano e pode ser tratado como uma variação do conhecido problema da mochila (Knapsack Problem).

2. Modelagem formal
Representação de uma solução
Uma solução candidata pode ser representada por uma lista de valores binários.
Cada posição representa um produto:

0 significa que o produto não foi escolhido;
1 significa que o produto foi escolhido.

Por exemplo:

[1, 0, 1, 1, 0]

significa que os produtos 1, 3 e 4 foram escolhidos.
Espaço de busca
Considerando n produtos, cada produto possui duas possibilidades: ser escolhido ou não ser escolhido.
Portanto, o número total de soluções possíveis é:
2ⁿ
Por exemplo, para 20 produtos existem:
2²⁰ = 1.048.576 soluções
possíveis.
Isso mostra que o espaço de busca cresce rapidamente conforme aumentamos a quantidade de produtos.
Função objetivo
O objetivo é maximizar a utilidade total dos produtos escolhidos.
Podemos representar a função objetivo por:
Maximizar:
U(x) = Σ utilidade_i × x_i
onde x_i indica se o produto i foi escolhido.
Ao mesmo tempo, o custo total deve ser calculado por:
C(x) = Σ preço_i × x_i
Restrições
A principal restrição é que o valor total das compras não pode ultrapassar o orçamento disponível:

C(x) ≤ orçamento

Além disso, cada produto só pode ser escolhido uma vez, portanto:

x_i ∈ {0,1}

Uma solução que ultrapasse o orçamento é considerada inviável.

3. Classificação do problema

Este problema pode ser considerado difícil, pois é uma variação do problema da mochila, que é conhecido por ser NP-difícil.

Uma maneira simples de entender isso é observar que, para n produtos, existem 2ⁿ combinações possíveis. Se tentássemos verificar todas as possibilidades, a quantidade de combinações aumentaria muito rapidamente.

Por exemplo:

10 produtos → 1.024 soluções;
20 produtos → 1.048.576 soluções;
30 produtos → 1.073.741.824 soluções.

Assim, para grandes quantidades de produtos, testar todas as possibilidades pode ser muito demorado. Existem algoritmos e técnicas de otimização que conseguem resolver casos específicos de forma mais eficiente, mas o problema geral não possui um método conhecido que encontre sempre a solução ótima em tempo polinomial.

4. Código em Python

O código abaixo cria uma solução aleatória, calcula seu custo e sua utilidade e verifica se ela respeita a restrição de orçamento.

import random

# Lista de produtos
produtos = [
    ("Arroz", 30, 10),
    ("Feijão", 12, 9),
    ("Leite", 8, 7),
    ("Carne", 45, 10),
    ("Frango", 25, 9),
    ("Macarrão", 7, 6),
    ("Ovos", 15, 8),
    ("Frutas", 20, 9),
    ("Verduras", 15, 8),
    ("Café", 18, 7)
]

orcamento = 100

# Gera uma solução aleatória
solucao = [random.randint(0, 1) for _ in produtos]

# Calcula custo e utilidade
custo_total = 0
utilidade_total = 0

for i in range(len(produtos)):
    if solucao[i] == 1:
        nome, preco, utilidade = produtos[i]
        custo_total += preco
        utilidade_total += utilidade

# Verifica se a solução é factível
factivel = custo_total <= orcamento

# Mostra os resultados
print("Solução:", solucao)
print("Produtos escolhidos:")

for i in range(len(produtos)):
    if solucao[i] == 1:
        print("-", produtos[i][0])

print("\nCusto total: R$", custo_total)
print("Utilidade total:", utilidade_total)

if factivel:
    print("Solução factível: respeita o orçamento.")
else:
    print("Solução não factível: ultrapassa o orçamento.")

5. Conclusão

O problema da lista de compras com orçamento limitado é um exemplo de problema de otimização combinatória. A solução precisa considerar simultaneamente os preços, as utilidades e o orçamento disponível.

A representação por valores binários facilita a modelagem, pois cada produto pode ser escolhido ou não. O objetivo é maximizar a utilidade total, enquanto a principal restrição é não ultrapassar o orçamento.

O código desenvolvido demonstra os principais conceitos da modelagem: geração de uma solução candidata, cálculo da função objetivo e verificação das restrições. Além disso, o crescimento do espaço de busca mostra por que esse tipo de problema pode se tornar difícil para uma quantidade grande de produtos.
```
