```python
import numpy as np

TAREFAS = np.array([12, 35, 40, 8, 15, 22, 19, 45, 60, 31, 14, 28, 50, 18, 25, 33, 42, 10, 5, 29])
NUM_TAREFAS = len(TAREFAS)
NUM_SERVIDORES = 4

TAM_POP = 100
GERACOES = 200
TAXA_MUTACAO = 0.1
USAR_ELITISMO = True

def calcular_makespan(individuo, tempos):
    cargas = np.zeros(NUM_SERVIDORES)
    for task_idx, server_idx in enumerate(individuo):
        cargas[server_idx] += tempos[task_idx]
    return np.max(cargas), cargas

np.random.seed(42)
populacao = [np.random.randint(0, NUM_SERVIDORES, size=NUM_TAREFAS) for _ in range(TAM_POP)]

for g in range(GERACOES):
    custos = [calcular_makespan(ind, TAREFAS)[0] for ind in populacao]
    melhor_idx = np.argmin(custos)

    novos = []
    if USAR_ELITISMO:
        novos.append(populacao[melhor_idx].copy())

    while len(novos) < TAM_POP:
        i1, i2 = np.random.choice(TAM_POP, 2, replace=False)
        pai1 = populacao[i1] if custos[i1] < custos[i2] else populacao[i2]

        i3, i4 = np.random.choice(TAM_POP, 2, replace=False)
        pai2 = populacao[i3] if custos[i3] < custos[i4] else populacao[i4]

        filho = pai1.copy()
        mask = np.random.rand(NUM_TAREFAS) < 0.5
        filho[mask] = pai2[mask]

        for idx in range(NUM_TAREFAS):
            if np.random.rand() < TAXA_MUTACAO:
                filho[idx] = np.random.randint(0, NUM_SERVIDORES)

        novos.append(filho)

    populacao = novos

custos_finais = [calcular_makespan(ind, TAREFAS)[0] for ind in populacao]
melhor_idx = np.argmin(custos_finais)
melhor_solucao = populacao[melhor_idx]
makespan_opt, cargas_opt = calcular_makespan(melhor_solucao, TAREFAS)

print(f"[Exercício 3] Makespan Mínimo Obtido: {makespan_opt:.2f} s")
print(f"Cargas acumuladas por Servidor (0 a 3): {cargas_opt}")
print(f"Vetor de Alocação (Servidor por Tarefa): {melhor_solucao}")
```
